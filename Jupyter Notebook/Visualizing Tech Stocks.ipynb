# Visualizing Tech Stocks

This analysis will review 5 technology stocks (Microsoft (MSFT), Amazon (AMZN), Apple (AAPL), Google (GOOGL), and Meta (META)) over a six month period (January 2023 through June 2023).  The libraries used will be pandas, numpy, yfinance, and Matplotlib.

```python
# Importing libraries
# %matplotlib inline requires Jupyter Notebook to plot the chart below the line of code instead of a popup

import pandas as pd
import numpy as np
import yfinance as yf
import matplotlib.pyplot as plt
%matplotlib inline  
```

```python
# creating variables to be used to pull the required stock data and pulling the data 

symbols = ["MSFT", "AMZN", "AAPL", "GOOGL", "META"]
start_date = "2023-01-01"
end_date = "2023-07-01"
stock_data = yf.download(symbols, start=start_date, end=end_date, auto_adjust=False)
stock_data
```

```python
# Isolating the data for adjusted close price to perform the analysis on this datapoint.  

adj_close = stock_data['Adj Close']
adj_close
```

```python
# Plotting the adjusted closing prices over time

adj_close.plot()
plt.xlabel('Date')
plt.ylabel('Adjusted Closing Price')
plt.title('Tech Stocks Adjusted Price Over Time')
plt.tight_layout()
plt.show()
```

```python
# Calculating and plotting the daily simple rate of return over time

stock_data_daily_returns = adj_close.pct_change()

plt.figure(figsize=(16,9))
stock_data_daily_returns.plot()
plt.xlabel('Date')
plt.ylabel('RoR')
plt.title('Daily Simple Rate of Return Over Time')
plt.tight_layout()
plt.show()
```

```python
# Creating multiple individual plots for their simple rate of return review.  This makes reviewing individual stock movement much easier.

fig = plt.figure(figsize=(15, 15))
ax1 = fig.add_subplot(321)
ax2 = fig.add_subplot(322)
ax3 = fig.add_subplot(323)
ax4 = fig.add_subplot(324)
ax5 = fig.add_subplot(325)
ax1.plot(adj_close["AAPL"].pct_change())
ax1.set_title("Apple")
ax2.plot(adj_close["META"].pct_change())
ax2.set_title("Facebook")
ax3.plot(adj_close["MSFT"].pct_change())
ax3.set_title("Microsoft")
ax4.plot(adj_close["AMZN"].pct_change())
ax4.set_title("Amazon")
ax5.plot(adj_close["GOOGL"].pct_change())
ax5.set_title("Google")
plt.tight_layout()
plt.show()
```

```python
# Calculating the mean of each tech stock's daily simple rate of return

daily_mean = stock_data_daily_returns.mean()
daily_mean
```

```python
# Getting the indexes for the x-axis

daily_mean.keys()
```

```python
# Getting the daily means for each stock for the y-axis

height =[]
for key in daily_mean.keys():
    height.append(daily_mean[key])
height
```

```python
# Arranging the keys on the x-axis based on the length

x_pos = np.arange(len(daily_mean.keys()))
x_pos
```

```python
# Plotting the means of each stock

plt.bar(x_pos, height)
plt.xticks(x_pos, daily_mean.keys())
plt.xlabel('Tech Stocks')
plt.ylabel('Daily Average')
plt.title('Daily Average Rate of Return')
plt.tight_layout()
plt.show()
```

Based on the visual above, Meta had the highest daily average rate of return.  If an investor is looking for higher rates of return, this stock would have been a good stock to invest in during the time period (first half of 2023).

```python
# Calculating the variance for each stock

daily_var = stock_data_daily_returns.var()
daily_var
```

```python
# Getting the index for the x-axis

daily_var.keys()
```

```python
# getting the individual variances to plot for comparison

height = []
for key in daily_var.keys():
    height.append(daily_var[key])
height
```

```python
# Arranging keys on the x-axis based on the length

x_pos = np.arange(len(daily_var.keys()))
x_pos
```

```python
# Plotting the variance for each stock

plt.bar(x_pos, height)
plt.xticks(x_pos, daily_var.keys())
plt.xlabel('Tech Stocks')
plt.ylabel('Variance')
plt.title('Daily Variance')
plt.tight_layout()
plt.show()
```

Based on the above visual, the stock with the least variance was Apple.  If an investor wanted a predictable return, Apple would have been a good choide at that time.  On the other hand, Meta had the largest variance, displaying that it could be a riskier stock to invest in at that time.  

```python
# Calculating the standard deviation

daily_std = stock_data_daily_returns.std()
daily_std
```

```python
# Getting the indexes for the x-axis to plot the data

daily_std.keys()
```

```python
# getting the bar heights for the standard deviation

height = []
for key in daily_std.keys():
    height.append(daily_std[key])
height
```

```python
# Arranging the keys on the x-axis based on the length

x_pos = np.arange(len(daily_std.keys()))
x_pos
```

```python
# Plotting the standard deviations

plt.bar(x_pos, height)
plt.xticks(x_pos, daily_std.keys())
plt.xlabel('Tech Stocks')
plt.ylabel('Standard Deviation')
plt.title('Daily Standard Deviation')
plt.tight_layout()
plt.show()
```

Based on the above visual, Meta has the highest standard deviation, indicating it is more volatile than the other stocks.  On the other hand, Apple has the lowest standard deviation indicating less volatility, a more predictable return stock.  

```python
# Calculating the correlations

stock_data_daily_returns.corr()
```

None of the stocks are negatively correlated, and should respond somewhat similarly to external economic factors.  All stocks are relatively close to each other regarding correlation, but some show a higher degree of correlation (such as Amazon and Microsoft, and Microsoft and Google) than others.
